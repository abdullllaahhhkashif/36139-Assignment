```python
raw.dropna(subset=["YIELD_KG"])
```

```python
raw["AREA_ACRES"].mean()
```

```python
raw["AREA_ACRES"]=raw["AREA_ACRES"].fillna(4.420168067226891)
raw
```

```python
raw["DISTRICT"]=raw["DISTRICT"].fillna("Unknown")
raw
```

```python
raw["FERTILIZER_USED"]=raw["FERTILIZER_USED"].fillna("No")
raw
```

```python
raw["YIELD_VALUE"]=raw["YIELD_VALUE"].fillna(raw["YIELD_VALUE"].mean())
raw
```

```python
raw["HARVEST_DATE"] = pd.to_datetime(raw["HARVEST_DATE"], errors="coerce")
raw
```

```python
raw["FARM_ID"] = raw["FARM_ID"].astype(str).str.replace("F", "")
raw["FARM_ID"] = raw["FARM_ID"].astype(int)
raw
```

```python
total_yield=raw.groupby("CROP_TYPE")["YIELD_KG"].sum()
total_yield
```

```python
import matplotlib.pyplot as plt
```

```python

plt.bar(total_yield.index,total_yield.values,color="red",width=0.75)
plt.xlabel("Crop Type")
plt.ylabel("Total Yield")
plt.title("TOTAL YIELD BY CROPS")
plt.show
```
