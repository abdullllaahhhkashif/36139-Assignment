```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\DELL\Downloads\teaching_prices_messy.csv")
```

```python
df=df.drop_duplicates()
```

```python
df['item'] = df['item'].str.strip().str.title()
```

```python
df['price'] = df['price'].astype(str).str.replace(',', '', regex=False)
```

```python
df['price'] = pd.to_numeric(df['price'], errors='coerce')
```

```python
df.loc[df['price'] == 9999, 'price'] = pd.NA
```

```python
summary_stats = df.groupby('item')['price'].agg(['mean', 'median', 'count', 'min', 'max'])
```

```python
summary_stats
```

```python
df['date'] = pd.to_datetime(df['date'])
df_sorted = df.sort_values(['item', 'city', 'date'])
```

```python
first_last = []
for item, group in df_sorted.groupby('item'):
    g_clean = group.dropna(subset=['price'])
    f_val = g_clean.iloc[0]['price']
    l_val = g_clean.iloc[-1]['price']
    pct_chg = ((l_val - f_val) / f_val) * 100
    first_last.append({'item': item, 'first_price': f_val, 'last_price': l_val, 'pct_change_%': pct_chg})
```

```python
print(pd.DataFrame(first_last))
```
